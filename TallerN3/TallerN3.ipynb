* Ejercicio 1. Completar Newton (Algoritmo 2.3), unos 20 minutos


```python
def newton_raphson(f, df, p0, TOL=1e-6, N0=50):
    """Algoritmo 2.3. Devuelve (p, historia), con historia = [p0, p1, ..., p]."""
    historia = [p0]
    i = 1
    while i <= N0:
        d = df(p0)
        if d == 0:
            raise ZeroDivisionError(f"f'(p0) = 0 en p0 = {p0}: la tangente es horizontal")
        p = p0 - f(p0) / d
        historia.append(p)
        if abs(p - p0) < TOL:
            return p, historia
        i = i + 1
        p0 = p
    raise RuntimeError(f"El método fracasó después de {N0} iteraciones")
    
```

* Ejercicio 2. Escribir la secante (Algoritmo 2.4), unos 45 minutos


```python
def secante(f, p0, p1, TOL=1e-6, N0=50):
    """Algoritmo 2.4. Devuelve (p, historia), con historia = [p0, p1, p2, ..., p]."""
    i = 2
    q0 = f(p0)
    q1 = f(p1)
    historia = [p0, p1]

    while i <= N0:
        if q1 == q0:
            raise ZeroDivisionError("La secante es horizontal (q1 == q0): división por cero")

        p = p1 - q1 * (p1 - p0) / (q1 - q0)
        historia.append(p)

        if abs(p - p1) < TOL:
            return p, historia
        i = i + 1

        p0, q0 = p1, q1
        p1 = p
        q1 = f(p)
    raise RuntimeError(f"El metodo fracaso después de {N0} iteraciones")
```

```python
llamadas = [] 

def f(x):
    llamadas.append(x)
    return x**3 + 4 * x**2 - 10


llamadas.clear()
p, historia = secante(f, 1.0, 2.0, TOL=1e-6)

print("Raiz p:", p)
print("Iteraciones:", len(historia) - 2)
print("Evaluaciones de f:", len(llamadas))
print("p2, p3, p4, p5:", historia[2:6])
```

* Ejercicio 3. ¿Qué tasa de interés cobra el banco? Unos 30 minutos


```python
def saldo_final(i, monto=20000.0, cuota=450.0, meses=60):
    """Saldo despues de pagar todas las cuotas, con interes mensual i (0.01 es 1 %)."""
    saldo = monto
    for mes in range(meses):
        saldo = saldo * (1 + i) - cuota
    return saldo
```

```python
print("saldo_final(0):", saldo_final(0))
i_mensual, historia_prestamo = secante(saldo_final, 0.005, 0.02, TOL=1e-10)
tea = (1 + i_mensual) ** 12 - 1

print(f"Tasa mensual i (%): {i_mensual * 100:.4f}")
print(f"TEA (%): {tea * 100:.2f}")
print("Saldo final en i_mensual:", saldo_final(i_mensual))
```
